```python
# importar librerías

# manipulación y transformación de datos
import pandas as pd
import numpy as np

# visualización
import matplotlib as plt
import matplotlib.pyplot as plt

# manipulación de fechas
import datetime

# ignorar advertencias
import warnings
warnings.filterwarnings('ignore')
```

```python
# # ------------------------------------------------------------------------
# # la siguiente línea permite crear un dataframe con las columnas declaradas dentro de los corchetes cuadrados dobles [[]], separados por comas, y cada nombre dentro de comillas en caso de ser formato texto
# df_casos = df_casos[['Tipo de caso PC','Descripción','A cargo','Viaje','Buque','Cliente ','Fecha de apertura','Unidad de Negocio','Ejecutivo','Estado','Resuelto por','Fecha de modificación','Fecha asigno caso','Fecha atendió Facturación','Fecha de resolución']]
# # ------------------------------------------------------------------------
```

```python
# creación de una variable nueva, únicamente se declara poniendo el nombre del dataframe y el nombre de la columna nueva dentro de corchetes cuadrados y comillas en caso de ser formato texto
# la función np.where permite seleccionar elementos según una condicional, la estructura es np.where(condicional, valor si verdadero, valor en otro caso)

# esta línea permite rellenar los valores nulos de la columna Fecha atendió Facturación con la Fecha de resolución menos 5 minutos
df_casos['Fecha atendió Facturación'] = np.where(df_casos['Fecha atendió Facturación'].isnull(),(df_casos['Fecha de resolución']-datetime.timedelta(minutes=5)),df_casos['Fecha atendió Facturación'])

# esta línea permite rellenar los valores nulos de la columna Fecha asigno caso con la Fecha atendió Facturación
df_casos['Fecha asigno caso'] = np.where((df_casos["Fecha asigno caso"].isnull()),df_casos['Fecha atendió Facturación'],df_casos['Fecha asigno caso'])
```

```python
df_casos.shape
```

```python
# para poder filtrar un dataframe se deben de ocupar los corchetes cuadrados y dentro de ellos debe de establecerse la condicional, la estructura es df[df['nombre de columna'].condicional] ej. df[df['columna'].isnull()] o df[df['columna'] > 30] o df[(df['columna_x'].isnull()) & (df['columna_y'] > 30)] o df[(df['columna_x'].isnull()) || (df['columna_y'] > 30)]

# en este caso, realizamos un filtro donde conservamos los registros donde la Fecha asigno caso no es nula
df_casos = df_casos[(df_casos['Fecha asigno caso'].notnull())]
```

```python
# imprimos las columnas que acabamos de crear
df_casos[['Fecha asigno caso', 'Fecha atendió Facturación', 'Fecha de resolución']].iloc[0]
```

```python
# # se creará una nueva columna, esta será calculada y será la diferencia entre la Fecha de resolución y la Fecha atendió Facturación, y se dividira entre la clase pd.Timedelta con el parametro de hours, ya que se busca obtener la diferencia en horas, pero esta clase puede ocuparse para obtener segundos, minutos, días, semanas, etc.

# # ------------------------------------------------------------------------
# # en caso de que se quiera obtener la diferencia en días
# df_casos['Tiempo VSC en días'] = ((df_casos['Fecha de resolución'] - df_casos['Fecha atendió Facturación']) / pd.Timedelta(days=1))
# # ------------------------------------------------------------------------

# en caso de que se quiera obtener la diferencia en horas
df_casos['Tiempo VSC en horas'] = ((df_casos['Fecha de resolución'] - df_casos['Fecha atendió Facturación']) / pd.Timedelta(hours=1))
```

```python
# # ------------------------------------------------------------------------
# # para renombrar una columna se ocupa el método rename donde el parámetro de columns debe de ir en formato de diccionario, entre corchetes y estructurado como un glosario, ej. df.rename(columns={'nombre_viejo': 'nombre_nuevo'})
# # OJO: esta línea no se ocupa ya que esta correción no es necesaria
# df_casos = df_casos.rename(columns={'Tiempo Facturación':'Tiempo VSC en horas'})
# # ------------------------------------------------------------------------
```

```python
# # de forma similar, se repetira el proceso de arriba para calcular la direncia entre Fecha atendió Facturación menos la Fecha de apertura

# # ------------------------------------------------------------------------
# # en caso de que se quiera obtener la diferencia en días
# df_casos['Tiempo Facturación en días'] = ((df_casos['Fecha atendió Facturación'] - df_casos['Fecha de apertura']) / pd.Timedelta(days=1))
# # ------------------------------------------------------------------------

# en caso de que se quiera obtener la diferencia en horas
df_casos['Tiempo Facturación en horas'] = ((df_casos['Fecha atendió Facturación'] - df_casos['Fecha de apertura']) / pd.Timedelta(hours=1))
```

```python
# # de forma similar, se repetira el proceso de arriba para calcular la direncia entre Fecha de resolución menos la Fecha de apertura

# # ------------------------------------------------------------------------
# # en caso de que se quiera obtener la diferencia en días
# df_casos['Tiempo Total en días'] = ((df_casos['Fecha de resolución'] - df_casos['Fecha de apertura']) / pd.Timedelta(days=1))
# # ------------------------------------------------------------------------

# en caso de que se quiera obtener la diferencia en horas
df_casos['Tiempo Total en horas'] = ((df_casos['Fecha de resolución'] - df_casos['Fecha de apertura']) / pd.Timedelta(hours=1))
```

```python
# imprimos las columnas que acabamos de crear
df_casos[['Tiempo VSC en horas', 'Tiempo Facturación en horas', 'Tiempo Total en horas']].iloc[0]
```

```python
# la siguiente línea permite obtener la semana de una columna de formato fecha, para poder hacer uso de esta se ocupara el accesorio 'dt' de pandas, el cual permite manipular elementos con formato de fecha, y el método de isocalendar el cual nos ayudará a obtener algún elemento especifico de una fecha según el estándar ISO8601, en este caso, estaremos obteniendo la semana
df_casos['Week']= df_casos['Fecha de resolución'].dt.isocalendar().week
```

```python
# imprimos las columnas que acabamos de crear
df_casos[['Fecha de resolución', 'Week']].iloc[0]
```

```python
# para imprimir algun valor o variable, solo se debe declarar, no es necesario ocupar la función print
df_casos.iloc[0]
```

```python
# debido a que se busca realizar análisis agrupados, se hará uso del método groupby(), este permite agrupar por columnas y a su vez se pueden ocupar diferentes funciones para obtener coeficientes, estos son: sum() para sumar los valores de la columna, min() y max() para obtener el valor mínimo y máximo de la columna,
# mean() para obtener el promedio de los valores de la columna, size() para obtener el tamaño de las columnas, count() para obtener un conteo de los valores de las columnas, etc.

# ------------------------------------------------------------------------
# son similares a los cálculos que se realizan con tablas dinámicas en excel
# ------------------------------------------------------------------------

# df_casos.groupby(['Week','Tipo de caso PC'])['No. de Folio'].size()[:-4]

# Es una prueba para sacar los casos de forma diaria
df_casos_diarios = df_casos.copy()
df_casos_diarios["Day"]=df_casos_diarios["Fecha de apertura"].dt.day
df_casos_diarios_agrup = df_casos_diarios.groupby(['Day','Tipo de caso PC'])['No. de Folio'].size().reset_index()
sum_casos = df_casos_diarios_agrup['No. de Folio'].sum()
max_casos = df_casos_diarios_agrup['No. de Folio'].max()
min_casos = df_casos_diarios_agrup['No. de Folio'].min()
mean_casos = df_casos_diarios_agrup['No. de Folio'].mean()
median_casos = df_casos_diarios_agrup['No. de Folio'].median()
```

```python
pd.DataFrame([{'Sum': int(sum_casos), 'Max': int(max_casos), 'Min': int(min_casos), 'Mean': round(mean_casos), 'Median': int(median_casos)}]).T
```

```python
# en este caso estaremos agrupando por semana y tipo de caso, y únicamente realizaremos un conteo del tamaño de los valores de la columna No. de Folio según la agrupación
# ocuparemos a su vez el método unstack() para poder generar un nuevo df con un nuevo nivel de columnas, ocuparemos de ejemplo la variable 'Tipo de caso PC', la cual tiene únicamente tres tipos de casos, los cuales son: CAMBIO DE BUQUE, CAMBIO DE POD, CAMBIO DE TRÁFICO

# # ------------------------------------------------------------------------
# # para obtener los valores únicos de una columna, se puede ocupar cualquiera de las siguientes siguiente líneas:
# df_casos.groupby(['Tipo de caso PC']).size() # conoces los únicos y se realiza un conteo
# df_casos['Tipo de caso PC'].unique() # solo conoces los únicos
# # ------------------------------------------------------------------------

# esta funcipon nos permitirá ocupar los valores de una columna como nombres de columnas, de esta forma, tendremos una columna para conocer la cantidad de casos por semana
df_casos.groupby(['Week','Tipo de caso PC'])['No. de Folio'].size()[:-4].unstack(level='Tipo de caso PC')
```

```python
# declaramos un nuevo dataframe para mantener el orden, este representará la cantidad de casos por semana, donde los tipos de casos serán columnas

# ------------------------------------------------------------------------
# OJO: el orden de los factores si altera el producto, las agrupaciones serán realizadas según el orden que se establezca de las columnas
# ------------------------------------------------------------------------

df_casos_tipo = df_casos.groupby(['Week','Tipo de caso PC'])['No. de Folio'].size()[:-4].unstack(level='Tipo de caso PC')
```

```python
# imprimoos las columnas que acabamos de crear
df_casos_tipo
```

```python
# para poder graficar se pueden utilizar diferentes propiedades, para este caso, usaremos la propiedad de plot, donde ajustaremos los parámetros de kind='bar' y stacked='True', indicando una gráfica de barras apiladas, el parámetro figsize nos indica el tamaño de la gráfica, donde figsize(ancho,alto), y color representará el color de nuestras p
# roporciones apiladas, se declararan en formato de lista, dentro de corchetes cuadrados [], separados por coma, y entre comillas
df_casos_tipo.plot(kind='bar', stacked=True, figsize=(15, 6), color=['#27B868', '#33C3FF','#3A27B8'])

# para ponerle título a una gráfica ocupamos la función title y el parámetro fontsize (que indica el tamaño de la letra) con el 'sufijo' plt, el cual es un modulo que permite editar los elementos de una gráfica
plt.title('Número de casos por Semana y Tipo', fontsize=14)

# agregamos el nombre del eje x con la función xlabel con el parámetro fontsize
plt.xlabel('Semana', fontsize=12)

# agregamos el nombre del eje y con la función ylabel con el parámetro fontsize
plt.ylabel('Casos', fontsize=12)

# agregamos la leyenda, que corresponderá a los nombres de nuestras columnas, además agregamos el parámetro title que es para indicar el título de la leyenda
plt.legend(title='Tipo de caso PC')

# la función xticks es para cambiar la orientación del eje x
plt.xticks(rotation=0)

# usaremos la función tight_layout para ajustar en automático el relleno (espacio interno) y espaciado (especio externo) de la gráfica
plt.tight_layout()

# para guardar un gráfica ocupamos la función savefig, sus parámetros son el nombre de la imagen (se debe de  poner la extensión del tipo de imagen que quieres guardar), y el dpi que indica la calidad de la imagen
plt.savefig('grafica_barras_apiladas.png', dpi=300)
```
