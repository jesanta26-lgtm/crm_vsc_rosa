```python
#JESD: Calculamos el tiempo de atención en horas, para esto ocuparemos la propiedad dt.total_seconds() la cual nos permite obtener el tiempo en segundos, posteriormente dividiremos entre 3600 para obtener el tiempo en horas
df_casos['Tiempo atención (Hrs)'] = (df_casos['Fecha de resolución']- df_casos['Fecha atendió Facturación']).dt.total_seconds()/3600

#JESD: Calculamos el dia de la semana en que se resolvió el caso, para esto ocuparemos el accesorio 'dt' de pandas, el cual permite manipular elementos con formato de fecha, y el método de isocalendar el cual nos ayudará a obtener algún elemento especifico de una fecha según el estándar ISO8601, en este caso, estaremos obteniendo el día de la semana
df_casos['Semana'] = (df_casos['Fecha de apertura'].dt.isocalendar().week)

#JESD: Obtenemos el dia de la semana en que se resolvió el caso, para esto ocuparemos el accesorio 'dt' de pandas, el cual permite manipular elementos con formato de fecha, y el método de isocalendar el cual nos ayudará a obtener algún elemento especifico de una fecha según el estándar ISO8601, en este caso, estaremos obteniendo el día de la semana
dias = {'Monday':'Lunes','Tuesday':'Martes','Wednesday':'Miércoles','Thursday':'Jueves','Friday':'Viernes','Saturday':'Sábado','Sunday':'Domingo'}
df_casos['Núm día semana'] = (df_casos['Fecha de apertura'].dt.dayofweek)
df_casos['Día semana'] = (df_casos['Fecha de apertura'].dt.day_name().replace(dias))

df_casos['Tiempo total caso (Hrs)'] = ((df_casos['Fecha de resolución']- df_casos['Fecha de apertura']).dt.total_seconds()/3600)
```

```python
# graficaremos la cantidad de casos recibidos por semana para esto ocuparemos un gráfico de barras apiladas (stacked bar chart), el cual nos permitirá observar la cantidad de casos recibidos por semana .

# Calculamos el número de casos por semana
casos_semana = (
    df_casos.groupby('Semana')
    .size()
)

# Estadísticas
promedio = casos_semana.mean()
mediana = casos_semana.median()

# Crear figura
fig, ax = plt.subplots(figsize=(12, 6))

# Crear barras
barras = ax.bar(
    casos_semana.index.astype(str),
    casos_semana.values,
    color='steelblue',
    edgecolor='black'
)

# Agregar etiquetas con el número de casos
for barra in barras:
    altura = barra.get_height()

    ax.text(
        barra.get_x() + barra.get_width() / 2,
        altura + 0.5,
        f'{int(altura)}',
        ha='center',
        va='bottom',
        fontsize=9,
        fontweight='bold'
    )

# Línea de promedio
ax.axhline(
    promedio,
    color='red',
    linestyle='--',
    linewidth=2,
    label=f'Promedio ({promedio:.1f})'
)

# Línea de mediana (opcional)
ax.axhline(
    mediana,
    color='green',
    linestyle=':',
    linewidth=2,
    label=f'Mediana ({mediana:.1f})'
)

# Configuración de la gráfica
ax.set_title(
    'Casos recibidos por semana',
    fontsize=16,
    fontweight='bold'
)

ax.set_xlabel('Semana del año', fontsize=12)
ax.set_ylabel('Número de casos', fontsize=12)

ax.grid(axis='y', linestyle='--', alpha=0.4)

ax.legend()

plt.tight_layout()

plt.show()

print(f'mediana de casos por semana: {mediana:.1f}')
```

```python
# Definir la capacidad instalada del PLANNING SERVICES
#   2 recursos
#   47 horas efectivas por semana cada uno

horas_semana = (
    df_casos
    .groupby('Semana')['Tiempo atención (Hrs)']
    .sum()
    .round(2)
)

capacidad_semanal = 94

capacidad = pd.DataFrame({
    'Horas consumidas': horas_semana
})

capacidad['Capacidad instalada'] = capacidad_semanal

capacidad['Capacidad disponible'] = (
    capacidad['Capacidad instalada']
    - capacidad['Horas consumidas']
).round(2)

capacidad['Utilización (%)'] = (
    capacidad['Horas consumidas']
    / capacidad['Capacidad instalada']
    * 100
).round(1)

capacidad
```

```python
#graficamos la capacidad instalada y la utilización semanal de la capacidad de Planning Services, para esto ocuparemos un gráfico de líneas (line plot), el cual nos permitirá observar la tendencia de horas consumidas por semana y compararlas con la capacidad instalada.

plt.figure(figsize=(14,6))

# Tendencia de horas consumidas
plt.plot(
    capacidad.index,
    capacidad['Horas consumidas'],
    marker='o',
    linewidth=3,
    label='Horas consumidas'
)

# Línea horizontal de capacidad instalada
plt.axhline(
    y=47,    # Capacidad instalada asumiendo 50% del tiempo de los 2 recursos (94 horas)
    color='red',
    linestyle='--',
    linewidth=2,
    label='Capacidad instalada (94 h)'
)

# Etiquetas
for semana, horas, porcentaje in zip(
        capacidad.index,
        capacidad['Horas consumidas'],
        capacidad['Utilización (%)']):

    plt.text(
        semana,
        horas + 2,
        f'{porcentaje:.1f}%',
        ha='center',
        fontsize=10,
        fontweight='bold'
    )

plt.title(
    'Utilización semanal de la capacidad de Planning Services',
    fontsize=16,
    fontweight='bold'
)

plt.xlabel('Semana')
plt.ylabel('Horas')

plt.xticks(capacidad.index)

plt.grid(axis='y', linestyle='--', alpha=0.3)

plt.legend()

plt.tight_layout()

plt.show()
```

# KPIs PLANNING SERVICES

```python
# Tiempo desde la apertura hasta la asignación
df_casos['Tiempo asignación (Hrs)'] = (
    df_casos['Fecha asigno caso']
    - df_casos['Fecha de apertura']
).dt.total_seconds() / 3600

# Tiempo desde la apertura hasta que Planning Services inicia la atención
df_casos['Tiempo espera atención (Hrs)'] = (
    df_casos['Fecha atendió Facturación']
    - df_casos['Fecha de apertura']
).dt.total_seconds() / 3600

# Tiempo desde la asignación hasta que inicia la atención
df_casos['Tiempo espera después asignación (Hrs)'] = (
    df_casos['Fecha atendió Facturación']
    - df_casos['Fecha asigno caso']
).dt.total_seconds() / 3600

# Variables temporales adicionales
df_casos['Mes'] = (
    df_casos['Fecha de apertura']
    .dt.to_period('M')
    .astype(str)
)

df_casos['Hora apertura'] = df_casos['Fecha de apertura'].dt.hour

```

```python
# Resumen general de KPIs

resumen_kpis = pd.DataFrame({
    'Indicador': [
        'Total de casos',
        'Tiempo promedio de atención',
        'Mediana de atención',
        'Tiempo promedio total del caso',
        'Tiempo promedio antes de iniciar atención',
        'Casos atendidos en 30 minutos',
        'Casos atendidos en 1 hora'
    ],
    'Resultado': [
        len(df_casos),
        df_casos['Tiempo atención (Hrs)'].mean(),
        df_casos['Tiempo atención (Hrs)'].median(),
        df_casos['Tiempo total caso (Hrs)'].mean(),
        df_casos['Tiempo espera atención (Hrs)'].mean(),
        (df_casos['Tiempo atención (Hrs)'] <= 0.5).mean() * 100,
        (df_casos['Tiempo atención (Hrs)'] <= 1).mean() * 100
    ]
})

resumen_kpis['Resultado'] = resumen_kpis['Resultado'].round(2)

resumen_kpis
```

```python
# Cumplimiento SLA de 30 minutos y una hora

total_casos = len(df_casos)

casos_sla_30 = (df_casos['Tiempo atención (Hrs)'] <= 0.5).sum()

casos_sla_1h = (df_casos['Tiempo atención (Hrs)'] <= 1).sum()

casos_fuera_sla_1h = (df_casos['Tiempo atención (Hrs)'] > 1).sum()

sla = pd.DataFrame({
    'Indicador': [
        'Cumple SLA 30 min',
        'Cumple SLA 1 hora',
        'Fuera de SLA 1 hora'
    ],
    'Casos': [
        casos_sla_30,
        casos_sla_1h,
        casos_fuera_sla_1h
    ]
})

sla['Porcentaje (%)'] = (sla['Casos'] / total_casos * 100).round(1)

sla
```

```python
# funcion para clasificar el cumplimiento del SLA según el tiempo de atención

def clasificar_sla(tiempo):
    if pd.isna(tiempo):
        return 'Sin información'
    elif tiempo <= 0.5:
        return 'Hasta 30 min'
    elif tiempo <= 1:
        return 'Entre 30 min y 1 h'
    else:
        return 'Más de 1 h'


df_casos['Cumplimiento SLA'] = (
    df_casos['Tiempo atención (Hrs)']
    .apply(clasificar_sla)
)

df_casos['Cumplimiento SLA'].value_counts()
```

```python
# SLA por dia de la semana

orden_dias = [
    'Lunes',
    'Martes',
    'Miércoles',
    'Jueves',
    'Viernes',
    'Sábado',
    'Domingo'
]

sla_dia = (
    df_casos
    .groupby('Día semana', observed=False)
    .agg(
        Casos=('Tiempo atención (Hrs)', 'size'),
        Tiempo_promedio=('Tiempo atención (Hrs)', 'mean'),
        Mediana=('Tiempo atención (Hrs)', 'median'),
        Cumple_30_min=(
            'Tiempo atención (Hrs)',
            lambda x: (x <= 0.5).mean() * 100
        ),
        Cumple_1_hora=(
            'Tiempo atención (Hrs)',
            lambda x: (x <= 1).mean() * 100
        )
    )
    .reindex(orden_dias)
    .round(2)
)

sla_dia
```

```python
#SLA por semana

sla_semana = (
    df_casos
    .groupby('Semana')
    .agg(
        Casos=('Tiempo atención (Hrs)', 'size'),
        Tiempo_promedio=('Tiempo atención (Hrs)', 'mean'),
        Mediana=('Tiempo atención (Hrs)', 'median'),
        Cumplimiento_30_min=(
            'Tiempo atención (Hrs)',
            lambda x: (x <= 0.5).mean() * 100
        ),
        Cumplimiento_1_hora=(
            'Tiempo atención (Hrs)',
            lambda x: (x <= 1).mean() * 100
        )
    )
    .round(2)
)

sla_semana
```

```python
# Tiempo de espera frente al tiempo de atención

espera_vs_atencion = pd.DataFrame({
    'Métrica': [
        'Tiempo promedio antes de iniciar',
        'Mediana antes de iniciar',
        'Tiempo promedio de atención',
        'Mediana de atención',
        'Tiempo promedio total del caso'
    ],
    'Horas': [
        df_casos['Tiempo espera atención (Hrs)'].mean(),
        df_casos['Tiempo espera atención (Hrs)'].median(),
        df_casos['Tiempo atención (Hrs)'].mean(),
        df_casos['Tiempo atención (Hrs)'].median(),
        df_casos['Tiempo total caso (Hrs)'].mean()
    ]
}).round(2)

espera_vs_atencion
```

```python
df_casos['Proporción atención (%)'] = (
    df_casos['Tiempo atención (Hrs)']
    / df_casos['Tiempo total caso (Hrs)']
    * 100
).round(1)

df_casos['Proporción espera (%)'] = (
    100 - df_casos['Proporción atención (%)']
).round(1)

df_casos[
    [
        'Tiempo total caso (Hrs)',
        'Tiempo atención (Hrs)',
        'Proporción atención (%)',
        'Proporción espera (%)'
    ]
].describe().round(2)
```

```python
# Capacidad y productvidad semanal 

capacidad_semanal_casos = 47

desempeno_semanal = (
    df_casos
    .groupby('Semana')
    .agg(
        Casos=('Tiempo atención (Hrs)', 'size'),
        Horas_consumidas=('Tiempo atención (Hrs)', 'sum'),
        Tiempo_promedio=('Tiempo atención (Hrs)', 'mean'),
        Mediana=('Tiempo atención (Hrs)', 'median')
    )
    .round(2)
)

desempeno_semanal['Capacidad casos (Hrs)'] = capacidad_semanal_casos

desempeno_semanal['Capacidad disponible (Hrs)'] = (
    desempeno_semanal['Capacidad casos (Hrs)']
    - desempeno_semanal['Horas_consumidas']
).round(2)

desempeno_semanal['Utilización (%)'] = (
    desempeno_semanal['Horas_consumidas']
    / desempeno_semanal['Capacidad casos (Hrs)']
    * 100
).round(1)

desempeno_semanal['Casos por hora consumida'] = (
    desempeno_semanal['Casos']
    / desempeno_semanal['Horas_consumidas']
).round(2)

desempeno_semanal
```

```python
# Productividad por recurso

df_casos[['A cargo', 'Resuelto por']].head()

productividad_recurso = (
    df_casos
    .groupby('Resuelto por')
    .agg(
        Casos_resueltos=('Tiempo atención (Hrs)', 'size'),
        Horas_consumidas=('Tiempo atención (Hrs)', 'sum'),
        Tiempo_promedio=('Tiempo atención (Hrs)', 'mean'),
        Mediana=('Tiempo atención (Hrs)', 'median'),
        Cumplimiento_30_min=(
            'Tiempo atención (Hrs)',
            lambda x: (x <= 0.5).mean() * 100
        ),
        Cumplimiento_1_hora=(
            'Tiempo atención (Hrs)',
            lambda x: (x <= 1).mean() * 100
        )
    )
    .round(2)
)

productividad_recurso['Casos por hora'] = (
    productividad_recurso['Casos_resueltos']
    / productividad_recurso['Horas_consumidas']
).round(2)

productividad_recurso.sort_values(
    'Casos_resueltos',
    ascending=False
)
```

```python
# Performance por tipo de caso 

performance_tipo = (
    df_casos
    .groupby('Tipo de caso PC')
    .agg(
        Casos=('Tiempo atención (Hrs)', 'size'),
        Horas_consumidas=('Tiempo atención (Hrs)', 'sum'),
        Tiempo_promedio=('Tiempo atención (Hrs)', 'mean'),
        Mediana=('Tiempo atención (Hrs)', 'median'),
        Tiempo_maximo=('Tiempo atención (Hrs)', 'max'),
        Cumplimiento_30_min=(
            'Tiempo atención (Hrs)',
            lambda x: (x <= 0.5).mean() * 100
        ),
        Cumplimiento_1_hora=(
            'Tiempo atención (Hrs)',
            lambda x: (x <= 1).mean() * 100
        )
    )
    .round(2)
    .sort_values('Casos', ascending=False)
)

performance_tipo
```

```python
# Pareto de tipo de casos

pareto = (
    df_casos['Tipo de caso PC']
    .value_counts()
    .rename_axis('Tipo de caso')
    .reset_index(name='Casos')
)

pareto['Porcentaje (%)'] = (
    pareto['Casos']
    / pareto['Casos'].sum()
    * 100
).round(1)

pareto['Porcentaje acumulado (%)'] = (
    pareto['Porcentaje (%)']
    .cumsum()
    .round(1)
)

pareto
```

```python
# Volumen y tiempo por unidad de negocio

performance_unidad = (
    df_casos
    .groupby('Unidad de Negocio')
    .agg(
        Casos=('Tiempo atención (Hrs)', 'size'),
        Horas_consumidas=('Tiempo atención (Hrs)', 'sum'),
        Tiempo_promedio=('Tiempo atención (Hrs)', 'mean'),
        Mediana=('Tiempo atención (Hrs)', 'median'),
        Cumplimiento_1_hora=(
            'Tiempo atención (Hrs)',
            lambda x: (x <= 1).mean() * 100
        )
    )
    .round(2)
    .sort_values('Casos', ascending=False)
)

performance_unidad
```
